```python
# Cài đặt thư viện pyngrok trước khi import
!pip install pyngrok --quiet

import os
import subprocess
import time
import requests
from pyngrok import ngrok
from kaggle_secrets import UserSecretsClient

# ==========================================
# 1. LẤY THÔNG TIN BẢO MẬT TỪ KAGGLE SECRETS
# ==========================================
user_secrets = UserSecretsClient()
try:
    NGROK_TOKEN = user_secrets.get_secret("NGROK_TOKEN_1")
    SSH_PUBLIC_KEY = user_secrets.get_secret("SSH_PUBLIC_KEY")
    print("✅ Tải Secrets từ Kaggle thành công!")
except Exception as e:
    print("❌ Thất bại: Vui lòng kiểm tra lại tên Label trong mục Secrets ở thanh bên phải.")
    raise e

# ==========================================
# 2. CÀI ĐẶT VÀ CẤU HÌNH SSH SERVER (MÃ HÓA)
# ==========================================
print("Đang cấu hình môi trường SSH an toàn...")
# Cập nhật và cài đặt OpenSSH Server
!apt-get update -qq && apt-get install -y -qq openssh-server > /dev/null

# Tạo thư mục lưu trữ khóa cấu hình SSH cho root
!mkdir -p /root/.ssh
!chmod 700 /root/.ssh

# Ghi Public Key được mã hóa từ máy local vào authorized_keys
with open("/root/.ssh/authorized_keys", "w") as f:
    f.write(SSH_PUBLIC_KEY)
!chmod 600 /root/.ssh/authorized_keys

# Cấu hình SSH Daemon: Khóa hoàn toàn đăng nhập bằng mật khẩu thường, chỉ nhận SSH Key
!sed -i 's/#PermitRootLogin prohibit-password/PermitRootLogin prohibit-password/' /etc/ssh/sshd_config
!sed -i 's/#PasswordAuthentication yes/PasswordAuthentication no/' /etc/ssh/sshd_config

# Đảm bảo thư mục thực thi của sshd tồn tại và khởi động dịch vụ
!mkdir -p /var/run/sshd
subprocess.Popen(["/usr/sbin/sshd", "-D"])
print("✅ SSH Server đã được kích hoạt ngầm.")

# ==========================================
# 3. KẾT NỐI NGROK TUNNEL (TCP PORT 22)
# ==========================================
print("Đang mở cổng an toàn qua Ngrok...")
ngrok.set_auth_token(NGROK_TOKEN)

# Mở đường hầm TCP trỏ vào cổng 22 (Cổng mặc định của SSH)
ssh_tunnel = ngrok.connect(22, "tcp")

# Định dạng lại chuỗi kết nối để bạn dễ dàng copy vào VS Code
raw_url = ssh_tunnel.public_url.replace("tcp://", "")
host, port = raw_url.split(":")

print("\n" + "="*60)
print("🔒 THIẾT LẬP KẾT NỐI SSH THÀNH CÔNG (ĐÃ MÃ HÓA KHÓA)")
print("="*60)
print(f"Địa chỉ Host (gõ vào VS Code): {host}")
print(f"Cổng Port  (gõ vào VS Code): {port}")
print("\n👉 Cú pháp lệnh SSH kết nối nhanh từ Terminal Local:")
print(f"ssh root@{host} -p {port} -i ~/.ssh/id_rsa_kaggle")
print("="*60)
```
